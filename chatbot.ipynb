```python
import ollama
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
```

```python

loader = TextLoader("PCB.txt")
docs = loader.load()
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=20)
splitdoc = text_splitter.split_documents(docs)
```

```python
pip install langchain-huggingface
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python

pip install sentence-transformers 
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = FAISS.from_documents(splitdoc, embeddings)
```

```python
pip install streamlit
```

```python
mock_camera_feed = {
    "image1.jpg": "FAIL: Solder bridge detected between microchip pins.",
    "image2.jpg": "FAIL: Resistor footprint is empty. Component is completely missing.",
    "image3.jpg": "PASS: All components aligned properly."
}
```

```python
test_file = "image2.jpg"
vision_finding = mock_camera_feed[test_file]
```

```python
matching_chunks = db.similarity_search(vision_finding, k=1)
rag_context = "\n".join([doc.page_content for doc in matching_chunks])
```

```python
gemma_prompt = f"""
You are a student Electronics Engineering Assistant.
Answer the question using ONLY the facts provided in the Context.

Context:
{rag_context}

Question:
The camera found this defect: '{vision_finding}'. 
What is the root cause and how do we fix the design layout?
"""
```

```python
!ollama pull gemma2
```

```python
print("Gemma 2 is analyzing...")
response = ollama.chat(
    model="gemma2",
    messages=[{"role": "user", "content": gemma_prompt}]
)

print("\n--- Engineering Remediation Report ---")
print(response['message']['content'])
```
